```python
from scipy import stats
```

```python
# Independent T-Test
yellow_taxi_tip = df[df['color']=='yellow']['tip']
green_taxi_tip = df[df['color']=='green']['tip']

t_stat, p_val = stats.ttest_ind(yellow_taxi_tip, green_taxi_tip)
print("Independent T-Test statistic:", t_stat)
print("P-value:", p_val)

if p_val < 0.05:
    print("Reject null hypothesis: tip of taxi customer is differ by color of taxi")
else:
    print("Fail to reject null hypothesis: No significant tip difference by color of taxi")

#if reject h0 means there is effect of color on tip in taxi
```

```python
card_pay=df[df['payment']=='credit card']['tip']
cash_pay=df[df['payment']=='cash']['tip']

t_stat_pay,p_val_pay=stats.ttest_ind(card_pay,cash_pay)
print("Independent T-Test statistic:", t_stat_pay)
print("P-value:", p_val_pay)

if(p_val_pay<0.05):
    print('reject null hypothesis : tip collected by taxi is differ by mode of payment(card or cash)')
else:
    print('fall to reject null hypothesis: No significant tip difference by payment mode')
```

```python
pickup_m=df[df['pickup_borough']=='Manhattan']['tip']
pickup_q=df[df['pickup_borough']=='Queens']['tip']
pickup_bx=df[df['pickup_borough']=='Bronx']['tip']
pickup_bn=df[df['pickup_borough']=='Brooklyn']['tip']
f_stat_pick,p_val_pick=stats.f_oneway(pickup_m,pickup_q,pickup_bx,pickup_bn)
print("ANOVA F-statistic:", f_stat_pick)
print("P-value:", p_val_pick)

if(p_val_pick<0.05):
    print('reject null hypothesis : tip of taxi customer is differ by pickup_borough of taxi')
else:
    print('fail to reject null hypothesis : No significant tip difference by pickup_borough')
```

### Hypotheses for correlation 
- H0 : correlation is zero
- H1 : correlation is non-zero


```python
r,p_val=stats.spearmanr(df['tip'],df['passengers'])
print("correlation : ",r)
print("P-value : ",p_val)
if(p_val<0.05):
    print('Reject null hypothesis : there is no relationship')
else:
    print('fail to reject null hypothesis')
```

```python
r,p_val=stats.pearsonr(df['tip'],df['fare'])
print("correlation : ",r)
print("P-value : ",p_val)
if(p_val<0.05):
    print('Reject null hypothesis : there is no relationship')
else:
    print('fail to reject null hypothesis')
```
