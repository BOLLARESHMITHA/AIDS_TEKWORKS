```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Social_Network_Ads.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X=df.iloc[:,[2,3]].values
y=df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier

base_model=DecisionTreeClassifier(max_depth=1)

```

```python
classifier=AdaBoostClassifier(estimator=base_model)
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,ConfusionMatrixDisplay
print('accuracy_score = ',accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
ConfusionMatrixDisplay.from_predictions(y_test,y_pred,cmap='Reds')
```

```python
classifier_gradient=GradientBoostingClassifier(
    n_estimators=100,
    random_state=42,
    max_depth=3
)
classifier_gradient.fit(X_train,y_train)
```

```python
y_pred_grad=classifier_gradient.predict(X_test)
```

```python
print('accuracy_score = ',accuracy_score(y_test,y_pred_grad))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_grad,
    cmap='Blues'
)
```

```python
from xgboost import XGBClassifier
```

```python
classifier_xg = XGBClassifier(
    n_estimators=100,      
    learning_rate=0.1,     
    max_depth=3,           
    subsample=0.8,         
    colsample_bytree=0.8,  
    random_state=42,
    eval_metric='logloss'
)

classifier_xg.fit(X_train, y_train)
```

```python
y_pred_xg = classifier_xg.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_xg))
print(classification_report(y_test, y_pred_xg))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_xg,
    cmap='Blues'
)
```
