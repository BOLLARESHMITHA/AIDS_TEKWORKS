```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Data
X = df[['enginesize']]
y = df['price']

# ---- Linear Regression ----
lin_model = LinearRegression()
lin_model.fit(X, y)

# ---- Polynomial Regression (degree 4) ----
poly = PolynomialFeatures(degree=5)
x_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(x_poly, y)

# ---- Sorting for smooth curves ----
x_sorted = np.sort(X,axis=0)

# Predictions
y_lin_pred = lin_model.predict(x_sorted)
y_poly_pred = poly_model.predict(poly.transform(x_sorted))

# ---- Plot ----
plt.scatter(X, y, color='steelblue')
plt.plot(x_sorted, y_lin_pred, color='blue', label='Linear Model')
plt.plot(x_sorted, y_poly_pred, color='orange', label='Polynomial Model')

plt.xlabel("Engine Size")
plt.ylabel("Car Price")
plt.title("Linear vs Curved Model Comparison")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score,root_mean_squared_error,mean_squared_error,mean_absolute_error

y_lin_pred = lin_model.predict(x_sorted)
y_poly_pred = poly_model.predict(poly.transform(x_sorted))
print('LinearRegression')
print('r2score = ',r2_score(y,y_poly_pred))
```
