```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
import matplotlib.pyplot as plt
```

```python
X=np.array([50,120,180,250,320,380]).reshape(-1,1)
y=np.array([500,1000,1500,1800,2500,2500])
```

```python
df_reg=DecisionTreeRegressor(
    max_depth=3,
    random_state=42
)
df_reg.fit(X,y)
```

```python
X_grid=np.linspace(0,400,200).reshape(-1,1)
y_pred=df_reg.predict(X_grid)
plt.scatter(X,y,color='blue')
plt.plot(X_grid,y_pred,color='red')
plt.show()
```

```python
n=int(input('enter no of units(consumed) : '))
print('this month eletricity bill = ',df_reg.predict([[n]])[0])
```

```python
X=np.array([1,2,3,4,5,7,8,9]).reshape(-1,1)
y=np.array([3,3.5,4,4.2,6,6.5,7,7.2])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.25)
```

```python
df_reg1=DecisionTreeRegressor(
    random_state=42
)
df_reg1.fit(X_train,y_train)
```

```python
X_grid=np.linspace(1,8,100).reshape(-1,1)
y_pred=df_reg1.predict(X_grid)
plt.scatter(X,y,color='blue')
plt.plot(X_grid,y_pred,color='red')
plt.show()
```

```python
df_deep=DecisionTreeRegressor(max_depth=None)
df_deep.fit(X_train,y_train)
```

```python
y_pred_deep=df_deep.predict(X_grid)
plt.scatter(X,y,color='blue')
plt.plot(X_grid,y_pred_deep,color='red')
plt.show()
```

```python
df_control=DecisionTreeRegressor(max_depth=3)
df_control.fit(X,y)
```

```python
y_pred_control=df_control.predict(X_grid)
plt.scatter(X,y,color='blue')
plt.plot(X_grid,y_pred_control,color='red')
plt.show()
```

```python
from sklearn.metrics import r2_score,mean_squared_error
y_test_pred_deep=df_deep.predict(X_test)
y_test_pred_con=df_control.predict(X_test)
print('non controlled')
print('r2_score = ',r2_score(y_test,y_test_pred_deep))
print('mean_squared_error = ',mean_squared_error(y_test,y_test_pred_deep))
print('controlled')
print('r2_score = ',r2_score(y_test,y_test_pred_con))
print('mean_squared_error = ',mean_squared_error(y_test,y_test_pred_con))
```

```python
np.random.seed(41)
X=np.arange(1,21).reshape(-1,1)
y=np.array([
    10,12,15,18,20,25,30,28,35,40,
    45,42,50,55,60,58,65,70,72,75
])
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
df_reg_sam=DecisionTreeRegressor(random_state=42)
df_reg_sam.fit(X_train,y_train)
```

```python
y_pred_sam=df_reg_sam.predict(X_test)
print('r2_score = ',r2_score(y_test,y_pred_sam))
print('mean_squared_error = ',mean_squared_error(y_test,y_pred_sam))
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
rf.fit(X_train,y_train)
```

```python
y_pred_rf=rf.predict(X_test)
print('r2_score = ',r2_score(y_test,y_pred_rf))
print('mean_squared_error = ',mean_squared_error(y_test,y_pred_rf))
```

```python
X_grid=np.linspace(1,20,100).reshape(-1,1)
plt.scatter(X,y)
plt.plot(X_grid,rf.predict(X_grid),color='green')
plt.plot(X_grid,df_reg_sam.predict(X_grid),color='red')
plt.show()
```
